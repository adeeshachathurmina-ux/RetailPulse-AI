## 1. Objective
The purpose of this notebook is to perform initial exploratory data profiling on the historical Walmart weekly sales dataset, assessing data types, completeness, date range integrity, and distributional properties.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set visual style
plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')
pd.set_option('display.float_format', lambda x: '%.2f' % x)

# Load raw dataset
data_path = '../data/Walmart_Store_sales.csv'
df = pd.read_csv(data_path)
df.columns = [c.strip().replace('\\_', '_') for c in df.columns]
df['Date'] = pd.to_datetime(df['Date'], dayfirst=True)
df = df.sort_values(['Store', 'Date']).reset_index(drop=True)

print(f"Dataset Shape: {df.shape[0]:,} rows, {df.shape[1]} columns")
df.head()
```

## 2. Structural & Quality Checks
Checking for missing values, duplicates, and column types to ensure reliable downstream modeling.

```python
print("--- Column Data Types ---")
print(df.dtypes)

print("\n--- Missing Values Count ---")
print(df.isna().sum())

print(f"\nDuplicate Rows: {df.duplicated().sum()}")
print(f"Unique Stores: {df['Store'].nunique()}")
print(f"Date Horizon: {df['Date'].min().strftime('%Y-%m-%d')} to {df['Date'].max().strftime('%Y-%m-%d')}")
```
